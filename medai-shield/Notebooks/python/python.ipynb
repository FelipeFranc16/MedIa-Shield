Operações basicas com funções.

```python
def soma(a,b):
    return a+b
def subtração(a,b):
    return a-b
def divisão(a,b):
    return a/b
def multiplicação(a,b):
    return a*b
def potencia(a,b):
    return a**b
def sobra(a,b):
    return a%b
def divint(a,b):
    return a//b
a = 2
b = 3
print('resultado da soma:', soma(a,b))
print('resultado da subtração:', subtração(a,b))
print('resultado da divisão:', divisão(a,b))
print('resultado da multiplicação:', multiplicação(a,b))
print('resultado da potencia:', potencia(a,b))
print('resultado da sobra:', sobra(a,b))
print('resultado da divisao com resultado inteiro:', divint(a,b))
```

Funções de string:
len()      : mostra o tamanho da string em numero de caracteres.
lower()    : transforma toda a string para caixa baixa.
upper()    : transforma toda a string para caixa alta.
str()      : converte em string.
isalpha()  : retorna false se a string contiver algum caracter que não seja letra.
isdigit()  : retorna false se a string contiver algum caracter que não seja númerico.
strip()    : remove os espaços em branco do começo e/ou final da string.
replace()  : substitui uma string por outra string.
split()    : devide a string em subtrings se encontrar ocorrencias do separador.

Testando essas funções.

```python
print('comprimento da string: ', len("inatel"))
print('Caixa baixa: ', "ABACAXI".lower())
print('Caixa alta: ', "abacaxi".upper())
print('pi: ', str(3.1415))
print('contém numeros?', "faca1345la".isalpha())
print('contém letras?', "5783232".isdigit())
print("      remove os espaços em branco.          ".strip())
print('Substituindo: ABACAXI'.replace('ABA','CASA'))
print('Hello, World!! 1, 2 ,3'.split(","))
```
